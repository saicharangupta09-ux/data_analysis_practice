## 2. Create a sample dataset
This example dataset represents student study hours, attendance, assignment scores, and exam scores. It intentionally includes a few data-quality issues for practice. The data is synthetic and is not a real student record.

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 108, 110],
    "name": ["Asha", "Rahul", "Meena", "Arjun", "Kiran", "Sara", "Vikram", "Neha", "Neha", "Imran"],
    "study_hours": [2.0, 3.5, 1.0, 5.0, 4.0, np.nan, 2.5, 6.0, 6.0, 3.0],
    "attendance_pct": [70, 82, 65, 95, 88, 75, 80, 98, 98, 105],
    "assignment_score": [65, 78, 55, 92, 85, 72, 74, 96, 96, 68],
    "exam_score": [60, 75, 50, 90, 82, 70, 73, 95, 95, 66],
    "department": ["AI&ML", "CSE", "AI&ML", "CSE", "ECE", "AI&ML", "ECE", "CSE", "CSE", "ai&ml"]
}
df = pd.DataFrame(data)
df
```

### 2.1 Inspect the dataset
Review the first and last rows, dimensions, column types, and summary statistics before making changes.

```python
print("First five rows:")
display(df.head())

print("Last three rows:")
display(df.tail(3))

print("Shape (rows, columns):", df.shape)
print("\nData types:")
print(df.dtypes)

print("\nNumeric summary:")
display(df.describe())

print("\nMissing values per column:")
print(df.isna().sum())
```

## 3. Pandas and NumPy operations
Practice selecting columns, filtering rows, sorting, grouping, aggregation, merging, transformation, and numerical calculations.

```python
# Select columns
display(df[["name", "study_hours", "exam_score"]].head())

# Filter rows using a condition
high_scores = df[df["exam_score"] >= 80]
display(high_scores[["name", "exam_score"]])

# Sort records by exam score (highest first)
display(df.sort_values("exam_score", ascending=False)[["name", "exam_score"]])

# Group by department and calculate aggregate statistics
display(
    df.groupby("department", dropna=False)
      .agg(
          average_exam_score=("exam_score", "mean"),
          maximum_assignment_score=("assignment_score", "max"),
          student_count=("student_id", "count")
      )
      .round(2)
)
```

```python
# NumPy numerical operations
scores = df["exam_score"].to_numpy(dtype=float)
print("Mean exam score:", np.nanmean(scores))
print("Median exam score:", np.nanmedian(scores))
print("Score standard deviation:", np.nanstd(scores, ddof=1))
print("Scores above or equal to 75:", scores[scores >= 75])

# Transformation: normalize exam scores to a 0–1 range
score_min = df["exam_score"].min()
score_max = df["exam_score"].max()
df["exam_score_normalized"] = (df["exam_score"] - score_min) / (score_max - score_min)
display(df[["name", "exam_score", "exam_score_normalized"]].head())
```

```python
# Merge example: student records and a small mentor lookup table
mentor_lookup = pd.DataFrame({
    "department": ["AI&ML", "CSE", "ECE"],
    "mentor": ["Dr. Rao", "Ms. Iyer", "Mr. Khan"]
})

# Standardize department labels before joining
df["department_clean"] = df["department"].str.strip().str.upper()
mentor_lookup["department_clean"] = mentor_lookup["department"].str.upper()

merged_df = df.merge(
    mentor_lookup[["department_clean", "mentor"]],
    on="department_clean",
    how="left",
    validate="many_to_one"
)
display(merged_df[["name", "department_clean", "mentor"]])
```

## 4. Data cleaning

```python
clean_df = df.drop(columns=["exam_score_normalized", "department_clean"], errors="ignore").copy()

# 1) Check duplicate rows and duplicate student IDs
print("Exact duplicate rows:", clean_df.duplicated().sum())
print("Duplicate student IDs:", clean_df["student_id"].duplicated().sum())

# Remove exact duplicate rows
clean_df = clean_df.drop_duplicates()

# 2) Standardize categorical text
clean_df["department"] = clean_df["department"].str.strip().str.upper()
clean_df["name"] = clean_df["name"].str.strip()

# 3) Convert numeric columns to numeric types safely
numeric_cols = ["study_hours", "attendance_pct", "assignment_score", "exam_score"]
for col in numeric_cols:
    clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

# 4) Validate ranges: attendance must be between 0 and 100
invalid_attendance = ~clean_df["attendance_pct"].between(0, 100)
print("Rows with invalid attendance:", int(invalid_attendance.sum()))
clean_df.loc[invalid_attendance, "attendance_pct"] = np.nan

# 5) Handle missing values
# Use the median for study hours and attendance; this is a simple teaching example.
clean_df["study_hours"] = clean_df["study_hours"].fillna(clean_df["study_hours"].median())
clean_df["attendance_pct"] = clean_df["attendance_pct"].fillna(clean_df["attendance_pct"].median())

print("\nMissing values after cleaning:")
print(clean_df.isna().sum())
print("\nCleaned dataset:")
display(clean_df)
```

## 5. Data visualizations
The following charts demonstrate category comparisons, distributions, relationships, and possible outliers.

```python
# Bar chart: average exam score by department
department_scores = clean_df.groupby("department")["exam_score"].mean().sort_values(ascending=False)

plt.figure(figsize=(7, 4))
department_scores.plot(kind="bar")
plt.title("Average Exam Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Exam Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Line chart: exam score by student ID (sorted for readability)
plot_df = clean_df.sort_values("student_id")

plt.figure(figsize=(8, 4))
plt.plot(plot_df["student_id"], plot_df["exam_score"], marker="o")
plt.title("Exam Score by Student ID")
plt.xlabel("Student ID")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()

# Note: IDs are identifiers, not time. This line chart is for practice only;
# a line chart is more meaningful when the x-axis has a natural order such as time.
```

```python
# Histogram: exam score distribution
plt.figure(figsize=(7, 4))
sns.histplot(data=clean_df, x="exam_score", bins=6, kde=True)
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python
# Scatter plot: study hours vs exam score
plt.figure(figsize=(7, 4))
sns.scatterplot(data=clean_df, x="study_hours", y="exam_score", hue="department", s=90)
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python
# Box plot: exam scores by department
plt.figure(figsize=(7, 4))
sns.boxplot(data=clean_df, x="department", y="exam_score")
plt.title("Exam Score Spread by Department")
plt.xlabel("Department")
plt.ylabel("Exam Score")
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap for numerical variables
numeric_df = clean_df[["study_hours", "attendance_pct", "assignment_score", "exam_score"]]
correlation_matrix = numeric_df.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Between Numerical Variables")
plt.tight_layout()
plt.show()

display(correlation_matrix.round(2))
```

Calculate descriptive statistics and inspect the relationships between variables. The observations below are prompts to validate against the outputs rather than conclusions to assume in advance.

## 7. Optional Scikit-learn practice
This section demonstrates a simple train/test split and linear regression model. It is included to verify that Scikit-learn is available and introduce a basic workflow. Because the dataset is tiny and synthetic, the evaluation is only illustrative.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

features = clean_df[["study_hours", "attendance_pct", "assignment_score"]]
target = clean_df["exam_score"]

X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.30, random_state=42
)

model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

print("Mean Absolute Error:", round(mean_absolute_error(y_test, predictions), 2))
print("Root Mean Squared Error:", round(np.sqrt(mean_squared_error(y_test, predictions)), 2))
print("R² score:", round(r2_score(y_test, predictions), 3))

display(pd.DataFrame({
    "Actual exam score": y_test.to_numpy(),
    "Predicted exam score": np.round(predictions, 2)
}).reset_index(drop=True))
```
